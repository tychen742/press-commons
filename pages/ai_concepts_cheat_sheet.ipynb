# AI Concepts Cheat Sheet

AI products are described with a small set of terms: LLM, RAG, MCP, skill, memory, and agent. The terms are easy to mix up because they name different kinds of things. One is a model, one is a technique, one is a protocol, and one is a whole system built from the others. This page defines each term in a few lines, says what kind of thing it is, and shows how the parts fit together.

Use this page as a reference while you read {{book_title}}. The chapters teach the ideas; this page keeps the vocabulary in one place. Product and model names change often, so this page uses none.

| Term | Kind of thing | What it does | Common mix-up |
| --- | --- | --- | --- |
| **LLM** (large language model) | Model | Takes text in and produces text out, one piece at a time. This lets it answer, summarize, write, and reason over what it is given. | It is not a database and not an agent. It knows only its training data and the text in the current request. |
| **Prompt** | Input | Everything sent to the model for one request: instructions, the conversation so far, and any added material. | The prompt is more than the question the user typed. |
| **Context window** | Limit | The most text, measured in tokens, that the model can read in one request. | It is not memory. Text left out of the window does not exist for the model. |
| **Embedding** | Data representation | A list of numbers that stands for the meaning of a piece of text. Texts with similar meanings get similar numbers. | An embedding model does not write text. It makes search by meaning possible. |
| **RAG** (retrieval-augmented generation) | Technique | Searches your documents for passages related to the question and adds them to the prompt before the model answers. | RAG does not retrain the model. It changes what the model reads. |
| **Fine-tuning** | Technique | Continues training a model on your examples so that its behavior changes. | It is rarely the right way to give a model current facts. |
| **Tool calling** | Mechanism | The model replies with a structured request to run a named function. A program runs the function and sends the result back. | The model never runs anything itself. |
| **MCP** (Model Context Protocol) | Protocol | A standard way for an AI application to connect to servers that offer tools and data. | MCP is a connector standard. It is not a tool, a model, or an agent. |
| **Skill** | Packaged instructions | A reusable set of instructions, sometimes with scripts and files, that the application loads when a task calls for it. | A skill is text the model reads at run time, not something it learned in training. |
| **Memory** | Stored context | Information saved outside the model and loaded into later prompts. | The model itself remembers nothing between requests. |
| **Agent** | System | A program that runs a model in a loop: the model chooses an action, the program carries it out with a tool, and the result goes back to the model until the task is done. | An agent is the whole system, not a kind of model. |
| **Guardrail** | Control | A check outside the model on inputs, outputs, or actions. | A sentence in the prompt asking the model to behave is an instruction, not a guardrail. |
| **Evaluation** | Practice | Measuring a system against a set of test cases with known expectations. | A demo that works once is not an evaluation. |

1. **Request.** A person or another program sends a request to the agent program.
2. **Load context.** The agent program gathers what the model needs to read: standing instructions, any skill that fits the task, and saved memory.
3. **Prompt.** The agent program sends all of it to the model as one prompt.
4. **Tool call or answer.** The model replies in one of two ways: a final answer, or a request to run a tool.
5. **Run tool.** The agent program, not the model, runs the tool. The tool may be a function in the program, a tool offered by an MCP server, or a search of your documents.
6. **Result.** The tool's result is added to the context, and steps 3 to 6 repeat until the model gives a final answer or the program reaches a limit.
7. **Answer.** The agent program returns the answer to the user.

The figure shows three things that a list of terms hides:

- **The model is one box.** Every other box is ordinary software that a developer writes, configures, and secures.
- **The agent is the whole Agent zone working with the other zones.** It is the model plus the program that loops, not a separate component beside them.
- **Everything the model knows about your organization arrives through context or tools.** Instructions, skills, memory, and retrieved documents are all text placed in the prompt.

Not every AI system uses every part. A chat assistant with no tools stops at step 4. A basic RAG system runs retrieval once, as a fixed step before the prompt, with no loop.

(commons-ai-model)=
## Model and Context

- **LLM.** A large language model is trained on a very large amount of text to predict what comes next. It works on **tokens**, which are words or pieces of words, and it writes its reply one token at a time.
- **Stateless.** Each request to a model stands alone. A conversation seems continuous only because the application sends the earlier messages again with every request.
- **Prompt.** The prompt is the full input for one request. It usually has a **system prompt** (standing instructions from the developer), the conversation so far, and any added material such as retrieved documents or tool results.
- **Context window.** The model can read only a limited number of tokens per request. When a conversation or document is too long, the application must shorten, summarize, or select what goes in.
- **Hallucination.** A model can write a fluent answer that is false, because it produces likely text, not checked facts. Giving it the right source text and asking it to cite that text reduces the problem but does not remove it.

(commons-ai-knowledge)=
## Knowledge: RAG and Fine-Tuning

A model's training data ends at a cutoff date and does not include your organization's private documents. **Retrieval-augmented generation (RAG)** closes that gap by looking things up at the time of the question.[^rag-source]

1. **Index (done ahead of time).** Split your documents into passages, compute an **embedding** for each passage, and store the embeddings in a search index.
2. **Retrieve.** Compute an embedding for the question and find the passages whose embeddings are closest to it.
3. **Augment.** Add those passages to the prompt, with their sources.
4. **Generate.** The model answers from the passages it was given.

There are three common ways to give a model knowledge, and they differ in cost and in how fast the knowledge can change.

| Approach | How it works | Best for | Limit |
| --- | --- | --- | --- |
| Put it in the prompt | Paste the material into the request. | A few short documents. | The context window, and the cost of sending the same text every time. |
| RAG | Retrieve only the relevant passages for each question. | Large or frequently changing document sets, and answers that must cite sources. | The answer is only as good as the passages retrieved. |
| Fine-tuning | Train the model further on your examples. | A consistent style, format, or specialized task. | Slow and costly to update, and it does not cite sources. |

(commons-ai-tools)=
## Tools and MCP

A model can only produce text. **Tool calling** (also called function calling) is the mechanism that lets that text cause an action.

1. The developer describes each tool to the model: a name, a description of what it does, and the inputs it takes.
2. When the model decides a tool is needed, it replies with the tool's name and input values instead of an answer.
3. The program checks the request, runs the tool, and sends the result back to the model.

The **Model Context Protocol (MCP)** standardizes the connection between AI applications and the systems that offer tools and data.[^mcp-source] An **MCP server** wraps a system such as a database, a file store, or a business application and lists what it offers. An AI application acts as an **MCP client** and connects to any number of servers. A connector written once as an MCP server then works with every application that supports the protocol.

- **MCP builds on tool calling; it does not replace it.** The model still asks for a tool by name, and the application still runs it. MCP defines how the application finds the tool and reaches it.
- **A tool is real access.** A tool that can send email or change a record can do so by mistake. Give each tool the least access the task needs.

(commons-ai-context)=
## Instructions, Skills, and Memory

Instructions, skills, and memory are all text that the application places in the prompt. They differ in who writes the text and when it is loaded.

| Kind of context | Who writes it | When it is loaded | Example |
| --- | --- | --- | --- |
| Instructions (system prompt) | The developer or administrator | Every request | "You are a support assistant. Answer only from company policy." |
| Skill | A developer or an experienced user | Only when the task matches | The steps and template for writing a weekly status report. |
| Memory | The application, as it runs | When it is relevant to a later request | "This customer prefers email to phone calls." |
| Retrieved documents | Your organization's own files | When a search finds them | The paragraph of the returns policy that answers the question. |

- **Skill.** A skill saves know-how once so that nobody has to explain the task again. Usually only a short description of each skill stays in the prompt, and the full instructions are loaded when needed, which saves space in the context window. The word comes from particular products; others use names such as custom instructions, rule files, or playbooks for the same idea.
- **Memory.** Short-term memory is the conversation history kept in the context window. Long-term memory is information the application saves to a file or database and loads again in a later session. Stored memory is data about people and their work, so it needs the same privacy and retention rules as any other record.

(commons-ai-agents)=
## Agents

An **agent** is a program that lets a model decide the next step, carries that step out with a tool, and repeats until the task is done. The pattern below shows the loop. It is a pattern, not runnable code.

```text
context = instructions + request
repeat until the step limit:
    reply = model(context)
    if reply is a final answer:
        return reply
    check the tool call against the guardrails
    result = run the tool
    context = context + reply + result
```

**Workflow or agent.** In a **workflow**, the developer fixes the steps in code and calls the model at some of them. In an agent, the model chooses the steps. A workflow is easier to predict, test, and audit; an agent handles tasks whose steps cannot be listed in advance. Use the simplest design that does the job.

Because an agent acts with less supervision, it needs controls that sit outside the model:

- **Step and cost limits** stop a loop that is not making progress.
- **Least-privilege tools** limit what a mistake can damage.
- **Human approval** is required before actions that are hard to undo, such as sending a message, paying, or deleting data.
- **Guardrails** check inputs, outputs, and tool calls against rules in code.
- **Logging** records each prompt, tool call, and result so that a failure can be traced.
- **Evaluation** runs the system against test cases before and after every change.
