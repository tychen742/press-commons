# Docker Cheat Sheet

Docker packages an application and everything it needs into an **image**, and runs that image as a **container**. The ten commands below cover the basic life cycle: get an image, run it, inspect it, stop it, and clean up. Run them in a terminal.

In the commands, `<container_id>` and `<image_id>` stand for the ID or name shown by `docker ps` and `docker images`.

## How the Commands Fit Together

1. `docker pull` (or `docker build`) puts an image on your computer.
2. `docker run` creates a container from an image and starts it.
3. `docker ps` shows what is running; `docker stop` and `docker start` pause and resume a container.
4. `docker rm` deletes a container, and `docker rmi` deletes an image. Stop and remove containers before removing the image they use.

A container is a running copy of an image, so removing a container does not remove its image.
